```python
#PS1_1

# 定义一个函数将a,b,c按从大到小进行排序
import random

def Print_values(a,b,c):
    
    # 判断a是否大于b
    if a > b:
        # 如果 a > b，继续判断 b > c
        if b > c:
            # 如果 b > c，按照 a、b、c 的顺序输出
            print(a,b,c)
        else:
            # 如果 a <= b，继续判断 a > c
            if a > c:
                # 如果 a > c，按照 a、c、b 的顺序输出
                print(a,c,b)
            else:
                # 如果 a <= c，按照 c、a、b 的顺序输出
                print(c,a,b)
    else:
        # 如果 a <= b，继续判断 b > c
        if b > c:
            # 如果 b > c，则不输出结果
            pass
        # 如果 b <= c，按照 c、b、a的顺序输出
        else:
            print(c,b,a)
          
print("开始测试随机值")
a = random.randint(1,100)
b = random.randint(1,100) 
c = random.randint(1,100)   

Print_values(a,b,c)
```

```python
#PS1_2

#2.1
import random

# 建立 M1 空矩阵
M1 = []

# 创建5行
for i in range(5):
    
    # 创建1行空列表
    row = []
    
    # 每行添加10个随机数值
    for j in range(10):
        
        number = random.randint(0,50)
        
        # 将随机数字添加到行中
        row.append(number)
    
    # 将行添加到 M1 中
    M1.append(row)
    
# 将 M1 矩阵结果输出
print(M1)
    

# 建立 M2(10,5) 矩阵
import random
M2 = []
for i in range(10):
    row = []
    for j in range(5):
        number = random.randint(0,50)
        row.append(number)
    M2.append(row)
    
print(M2)
    

# 2.2
def Matrix_multip(M1,M2):
    
    # 定义结果为result
    result = []
    
    # 使用 for 循环，重复生成至第5行
    for i in range(5):
        
        row = []
        
        # 使用 for 循环，重复生成至第5位元素
        for j in range(5):
            
            # 将当前矩阵位置的计算结果定义为 total,初始值为0
            total = 0
            
            # M1 有10列， M2 有10行，k 对应M1当前行的第 k 个元素 和M2当前列的第 k 个元素 
            for k in range(10):
                
                # 根据计算规律可得， M1ij 中 j = k, M2ij 中 i = k
                total = total + M1[i][k]* M2[k][j]
           
            row.append(total)
            
        result.append(row)
         
    return result

result = Matrix_multip(M1,M2)

print(result)
```

```python
#PS1_3

def Pascal_triangle(k):
    
    # 创建1个包含1的列表作为 Pascal_triangle 的第一行
    row = [1]#第 0 行
    # 使用 for 循环重复生成至 k 行,
    for i in range(k):
        
        #每次循环前创建一行包含1的新列表
        new_row = [1]
        
        # 使用 for 循环计算至上一行（元素数-1） 的次数 
        for j in range(len(row)-1):
            
            # 把上一行中两个相邻的元素相加所得的结果，添加至新的一行
            new_row.append(row[j] + row[j+1])
        
        # 在新一行的末尾加1
        new_row.append(1)
        
        # 将新生成的结果返还至 row 
        row = new_row
        
    print(row)
    
    
Pascal_triangle(100)
Pascal_triangle(200)
```

```python
#PS1_4
# 设目标金额参数为x
import random
x = random.randint(1,100)

print("Target amount",x)

def Least_moves(x):
    
    # 初始金额为1
    money = 1
    
    # 如果目标金额是1，程序停止
    if x == 1:
        return 0
    
    else:
        moves = Least_moves(x - 1) + 1
        
        # 判断条件
        if x % 2 == 0:
            moves = min(moves, Least_moves(x // 2)+1)
        
        # 将计算的最少步数返回 moves 
        return moves

print("Least moves", Least_moves(x))
```

```python
# PS1_5

def Find_expression(target):
    
    count = 0

    # 我的同学 Bian Yue 给我讲解了如何利用循环遍历所有可能的表达式，以及通过三进制表示不同的选择
    # 在1-9里面有8个空位来填写 + 或 - 或者数字链接，故有3^8个组合，将其全部遍历
    for i in range(3**8):
        expression = "1"
        number = i
        
        for digit in range(2,10):
            
            # 每个数字的三进制恰好表示一种可能
            choice = number % 3
            number = number // 3
            
            #通过基数转换算法提取出对应位的数字，进而转换为+ 或- 或者数字链接
            if choice == 0:
                expression = expression + str(digit)
            
            elif choice == 1:
                expression = expression + "+" +str(digit)
                
            else:
                expression = expression + "-" + str(digit)
                
        if eval(expression) == target:
            print(expression, "=", target)
            count = count + 1  
    #返回满足条件的表达式数量    
    return count 
    

#5.1
#调用函数
Find_expression(50)

#5.2
#设置列表保存满足条件的表达式的数量
Total_solutions = []

#从1到100 调用函数
for i in range(1,101):
    count = Find_expression(i)
    
    #把函数的返回值即满足条件的表达式的数量存入列表
    Total_solutions.append(count)

print("Total_solutions =", Total_solutions)#

#输出最多的
print("Maximum number of solutions:", max(Total_solutions))
#最少的
print("Minimum number of solutions:", min(Total_solutions))

print("Numbers with maximum solutions:")
for i in range(100):
    if Total_solutions[i] == max(Total_solutions):
        print(i + 1)

print("Numbers with minimum solutions:")
for i in range(100):
    if Total_solutions[i] == min(Total_solutions):
        print(i+1)
```
